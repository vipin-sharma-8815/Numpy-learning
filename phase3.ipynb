```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Data structure: [restaurant_id, 2021, 2022, 2023, 2024]

sales_data = np.array([
    [1, 150000, 180000, 220000, 250000],  # Paradise Biryani
    [2, 120000, 140000, 160000, 190000],  # Beijing Bites
    [3, 200000, 230000, 260000, 300000],  # Pizza Hub
    [4, 180000, 210000, 240000, 270000],  # Burger Point
    [5, 160000, 185000, 205000, 230000]   # Chai Point
])

print("-----Zomato sales Annalysis-----")

print("\n Sales data Shape:", np.shape(sales_data))

print("\nSample data for first three years:\n", sales_data[:3])
```

```python
# total sales per year

print(np.sum(sales_data, axis=0))
yeary_total = np.sum(sales_data[:,1:], axis=0)
print(yeary_total)

print("\nTotal sale is:", np.sum(yeary_total));


# Minimum sales per year

min_sales = np.min(sales_data[:,1:], axis=1)
print("\n",min_sales)


# Maximu sale per year
max_sales = np.max(sales_data[:,1:], axis=0)
print("\n", max_sales)

# Average sales per Resturant

avg_sales = np.mean(sales_data[:,1:], axis=1)
print("\n", avg_sales) 
```

```python
cumsum = np.cumsum(sales_data[:, 1:], axis=1)
print(cumsum)

plt.figure(figsize=(8, 6))
plt.plot(np.mean(cumsum, axis=0))
plt.title("Average cumulative sales across all restaurant")
plt.xlabel("Years")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

```python
vector1 = np.array([ 1, 2, 3, 4, 5])
vector2 = np.array([ 6, 7, 8, 9, 10])

print("Vector addition:", vector1 + vector2)

print("\nVector Multiplication", vector1*vector2)

print('\nDot Product:', np.dot(vector1, vector2))

angle = np.arccos(
    np.dot(vector1, vector2) /
    (np.linalg.norm(vector1) * np.linalg.norm(vector2))
)

print("\nAngle b/w vetors:",angle)
```
